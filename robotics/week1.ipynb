```python
import numpy as np

a = np.array([[1, 2, 3]])
b = np.array([[4, 5, 6]])

print("sum is : ", np.add(a, b))
print("dot product is : ", np.dot(a, b.T))
print("length of a is : ", np.linalg.norm(a))

assert np.allclose(np.add(a, b), np.array([[5, 7, 9]])), "Addition result is incorrect"
assert np.allclose(np.dot(a, b.T), np.array([[32]])), "Dot product result is incorrect"
assert np.isclose(np.linalg.norm(a), 3.7416573867739413), "Length of a is incorrect"
```
