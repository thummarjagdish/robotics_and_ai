```python
query = "How to reduce heart disease risk?"

documents = [
    "Eating fiber reduces heart disease risk.",
    "Fruits and vegetables lower cardiovascular disease risk.",
    "Buying a new car improves driving comfort.",
    "Regular exercise improves heart health."
]
```

```python
query_embedding = google_embeddeding_model.embed_query(query)
```

```python
document_embeddings = google_embeddeding_model.embed_documents(documents)

print("Query Embedding:", query_embedding)
print("Document Embeddings:", document_embeddings)

print("Query Embedding Length:", len(query_embedding))
print("Document Embeddings Lengths:", [len(doc_emb) for doc_emb in document_embeddings])
```

```python
import numpy as np

document_embeddings = google_embeddeding_model.embed_documents(documents)

query_vector = np.asarray(query_embedding)
document_vectors = np.asarray(document_embeddings)

similarities = (
    document_vectors @ query_vector
    / (np.linalg.norm(document_vectors, axis=1) * np.linalg.norm(query_vector))
)

ranked_results = sorted(
    zip(documents, similarities),
    key=lambda result: result[1],
    reverse=True,
)

for document, similarity in ranked_results:
    print(f"{similarity:.4f}  {document}")
```

```python
# Cosine similarity
cosine_scores = (
    document_vectors @ query_vector
    / (np.linalg.norm(document_vectors, axis=1) * np.linalg.norm(query_vector))
)

print("Ranked by cosine similarity:")
for index in np.argsort(cosine_scores)[::-1]:
    print(f"{cosine_scores[index]:.4f}  {documents[index]}")

# Dot product
dot_scores = document_vectors @ query_vector

print("\nRanked by dot product:")
for index in np.argsort(dot_scores)[::-1]:
    print(f"{dot_scores[index]:.4f}  {documents[index]}")
```
