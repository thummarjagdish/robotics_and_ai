TF-IDF practical


```python
from sklearn.feature_extraction.text import TfidfVectorizer 
```

```python
documents = [
    "people watch movie and watch movie again",
    "people watch cricket and watch cricket",
    "people like movie and like movie a lot",
    "people like cricket"]
```

```python
tf_idf_vectorizer = TfidfVectorizer()
```

```python
tf_idf_matrix = tf_idf_vectorizer.fit_transform(documents)

tf_idf_matrix.toarray()
```

```python
for vector in tf_idf_matrix.toarray():
    print(vector)
```

```python
print(tf_idf_vectorizer.get_feature_names_out())
```
