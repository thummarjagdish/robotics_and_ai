```python
sentance = "This is jagdish"

words = sentance.lower().split()

print(words )
```

```python
word_vectors = []

for word in words:
    if word in model:
        word_vectors.append(model[word])
    else:
        print(f"Word '{word}' not found in the model.")
```

```python
print(word_vectors)
print(len(word_vectors))
```

```python
import numpy as np
sentance_vector = np.mean(word_vectors, axis=0)
print(sentance_vector)
print(len(sentance_vector))
print(sentance_vector.shape)
```

```python
model.most_similar([sentance_vector], topn=10)
```
